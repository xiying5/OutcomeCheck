```python
import json
import time
```

```python
# Create a function that removes duplicates in the data
def process_nctid(nctid):

    ## Task 1: Parse elements (prospective) ##
    out_link=f"out_pro/{nctid}.json"

    # Remove duplicates for the same outcome
    # Load the JSON data
    with open(out_link, 'r') as file:
        data = json.load(file)

    # Remove duplicates while preserving order
    seen = set()
    unique_data = []
    for item in data:
        item_tuple = tuple(item.items())  # Convert dict to hashable type
        if item_tuple not in seen:
            seen.add(item_tuple)
            unique_data.append(item)

    # Save the updated data back to a JSON file
    with open(f'out_pro/{nctid}_merged.json', 'w') as file:
        json.dump(unique_data, file, indent=2)
```

```python
# Read NCT IDs
with open('nctid_new150.txt', 'r') as file:
    nctids = file.read().splitlines()

# Process each NCTID
# Initialize a counter variable
n = 1

for nctid in nctids:
    print(f"Processing NCTID {n}/{len(nctids)}: {nctid}")
    process_nctid(nctid)
    print(f"Finished processing NCTID: {nctid}")
    
    # Optional: Add a delay between processing each NCTID to avoid rate limiting
    time.sleep(5)
    
    # Increment the counter
    n += 1

print("All NCTIDs processed.")
```
